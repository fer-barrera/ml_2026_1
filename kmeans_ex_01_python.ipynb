# K-means en 2D: dos distribuciones separadas

Este notebook construye un ejemplo introductorio de agrupamiento con **K-means** usando datos bidimensionales simulados. El objetivo es observar cómo el algoritmo identifica dos grupos cuando las distribuciones están relativamente separadas.

La idea central es: no se entregan etiquetas al modelo; el algoritmo debe encontrar una estructura de agrupamiento a partir de la geometría de los datos.

## 1. Idea matemática del método

Cada muestra se representa como un punto:

$$
\mathbf{x}_i = [x_{i1}, x_{i2}] \in \mathbb{R}^2
$$

K-means busca $K$ centroides. Si $K=2$, el algoritmo intenta encontrar dos puntos representativos:

$$
\boldsymbol{\mu}_1, \boldsymbol{\mu}_2
$$

El criterio que minimiza es la suma de distancias cuadráticas entre cada muestra y el centroide del cluster asignado:

$$
J = \sum_{i=1}^{n} \left\|\mathbf{x}_i - \boldsymbol{\mu}_{z_i}\right\|^2
$$

Donde $z_i$ indica el cluster asignado a la muestra $i$. En términos prácticos, cada punto se asigna al centroide más cercano y luego los centroides se recalculan como el promedio de sus puntos.

## 2. Librerías

Usaremos `numpy` para generar los datos, `pandas` para inspeccionarlos, `matplotlib` para graficar y `KMeans` de `scikit-learn` para aplicar el algoritmo.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans

RANDOM_STATE = 7
rng = np.random.default_rng(RANDOM_STATE)

plt.rcParams["figure.figsize"] = (7, 5)
plt.rcParams["axes.grid"] = True
plt.rcParams["grid.alpha"] = 0.25
```

## 3. Generación de datos simulados

Se crean dos grupos de datos usando distribuciones gaussianas multivariadas. Cada distribución tiene:

- un vector de medias, que ubica el centro aproximado del grupo;
- una matriz de covarianza, que controla la dispersión y la orientación de la nube de puntos.

```python
mu1 = np.array([2, 3])
sigma1 = np.array([[1, 0.5],
                   [0.5, 1]])

mu2 = np.array([6, 8])
sigma2 = np.array([[1, -0.5],
                   [-0.5, 1.5]])

n_points = 100

data1 = rng.multivariate_normal(mu1, sigma1, n_points)
data2 = rng.multivariate_normal(mu2, sigma2, n_points)

data = np.vstack([data1, data2])

df = pd.DataFrame(data, columns=["dimension_1", "dimension_2"])
df.head()
```

## 4. Visualización de los datos sin etiquetas

En esta primera gráfica todos los puntos se muestran con el mismo estilo porque el algoritmo todavía no ha asignado ningún cluster. Esta es la situación típica del aprendizaje no supervisado: tenemos datos, pero no tenemos etiquetas conocidas para entrenar un clasificador.

```python
plt.figure()
plt.scatter(data[:, 0], data[:, 1], s=40, alpha=0.8)
plt.title("Datos 2D sin etiquetar")
plt.xlabel("Dimensión 1")
plt.ylabel("Dimensión 2")
plt.show()
```

## 5. Aplicación de K-means

Como se simularon dos distribuciones, se fija:

$$
K = 2
$$

El modelo ajusta los centroides y asigna cada punto al centroide más cercano.

```python
num_clusters = 2

kmeans = KMeans(n_clusters=num_clusters, random_state=RANDOM_STATE, n_init=10)
cluster_idx = kmeans.fit_predict(data)
cluster_centers = kmeans.cluster_centers_

cluster_centers
```

## 6. Visualización de los clusters encontrados

Los colores representan los clusters asignados por K-means. Las estrellas negras representan los centroides finales.

Es importante notar que los nombres de los clusters no tienen significado semántico. El cluster 0 o el cluster 1 son solo identificadores asignados automáticamente por el algoritmo.

```python
plt.figure()
plt.scatter(data[:, 0], data[:, 1], c=cluster_idx, cmap="tab10", s=40, alpha=0.85)
plt.scatter(cluster_centers[:, 0], cluster_centers[:, 1], marker="*", s=300, c="black", edgecolor="white", label="Centroides")
plt.title("Datos 2D agrupados con K-means")
plt.xlabel("Dimensión 1")
plt.ylabel("Dimensión 2")
plt.legend()
plt.show()
```
